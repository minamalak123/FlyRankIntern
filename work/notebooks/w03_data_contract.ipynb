### Data contract

**1. What one row means:**  
One row represents the daily performance of one anonymized content item for one anonymized client on one reporting date.

**2. Tables used:**  
The primary table is `fact_content_daily_performance`. I use its March 2026 partition for development. Content and client dimension tables may be used later only when needed for contextual joins.

**3. Time window:**  
Development uses the middle-panel month March 2026, from `2026-03-01` through `2026-03-31`. Any future outcome window must be kept separate from the feature window.

**4. What I predict/rank:**  
My lane is Refresh / Content Opportunity Scoring. The goal is to rank content items by a Refresh Opportunity Score proxy so editors can review higher-priority pages first. The proxy must be defined from measurable warehouse outcomes rather than treated as an existing label.

**5. Deliberate exclusion:**  
Client and content identifiers are deliberately excluded as predictive features because they identify entities rather than describe transferable performance patterns, and they can encourage memorization or entity leakage.

```python
con.execute(f"""
SELECT
    report_date,
    client_hash_id,
    content_hash_id,
    COUNT(*) AS duplicate_rows
FROM read_parquet('{march_file}')
GROUP BY
    report_date,
    client_hash_id,
    content_hash_id
HAVING COUNT(*) > 1
LIMIT 5
""").fetchdf()
```

## Verification 2 — Row count and date span

I measure the number of rows and the minimum and maximum `report_date` in the March 2026 development partition.

```python
con.execute(f"""
SELECT
    COUNT(*) AS row_count,
    MIN(report_date) AS first_date,
    MAX(report_date) AS last_date
FROM read_parquet('{march_file}')
""").fetchdf()
```

## 2. Fields: feature / label / context / excluded

My Week 2 lane is **Refresh / Content Opportunity Scoring**, framed as a ranking task.

The final model will rank webpages by a **Refresh Opportunity Score** proxy. The score is intended for decision support: content editors review high-ranked pages first.

Because the warehouse daily table is at daily × client × content grain, the final feature vector must be aggregated to the webpage/content decision unit before modeling.

At this stage, fields are classified as follows:

### Context
- `client_hash_id` — identifies the anonymized client and is used for grouping, not as a predictive feature.
- `content_hash_id` — identifies the anonymized content item and is used for grouping, not as a predictive feature.
- `report_date` — used to define the observation window and temporal ordering.
- `month` — used to select the development partition.

### Candidate features
- `gsc_impressions`
- `gsc_clicks`
- `gsc_avg_position`
- `ga4_pageviews`
- `ga4_sessions`
- `ga4_engaged_sessions`
- `ga4_total_engagement_sec`
- `sessions_organic`
- `sessions_direct`
- `sessions_referral`
- `sessions_social`
- `sessions_paid`
- `sessions_ai`
- `scroll_events`

These are candidate performance signals only. Each one must be checked against the prediction timing and leakage rules before it becomes a final feature.

### Availability/context flags
- `client_has_gsc`
- `client_has_ga4`
- `gsc_data_available`
- `ga4_data_available`

These will be used to distinguish unavailable source data from genuine zero measurements and will not automatically be treated as predictive features.

### Label / proxy
The ranking target is a **Refresh Opportunity Score** proxy. Its exact construction and prediction window must be defined before the final feature vector is built.

### Excluded
- `client_hash_id` — excluded as a predictive feature because it is an identifier and can encourage memorization/entity leakage.
- `content_hash_id` — excluded as a predictive feature because it is an identifier and should be used only for grouping/auditing.
- Any field derived from the Refresh Opportunity Score — excluded from the feature vector because it would leak the target.
- Any future-window field — excluded because it would not be available at the decision moment.

```python
# Show the fields currently classified for the contract.
field_groups = {
    "context": [
        "client_hash_id",
        "content_hash_id",
        "report_date",
        "month",
    ],
    "candidate_features": [
        "gsc_impressions",
        "gsc_clicks",
        "gsc_avg_position",
        "ga4_pageviews",
        "ga4_sessions",
        "ga4_engaged_sessions",
        "ga4_total_engagement_sec",
        "sessions_organic",
        "sessions_direct",
        "sessions_referral",
        "sessions_social",
        "sessions_paid",
        "sessions_ai",
        "scroll_events",
    ],
    "availability_context": [
        "client_has_gsc",
        "client_has_ga4",
        "gsc_data_available",
        "ga4_data_available",
    ],
    "excluded_identifiers": [
        "client_hash_id",
        "content_hash_id",
    ],
}

for group, columns in field_groups.items():
    print(f"\n{group.upper()}")
    for column in columns:
        print(f"  - {column}")
```

## 3. Verify it with queries (grain, counts, missing values, windows)

*Every claim above gets a query cell here. A contract claim without a query next to it is a guess.*

```python
# Verification 1 — Grain
grain_check = con.execute(f"""
SELECT
    COUNT(*) AS duplicate_groups
FROM (
    SELECT
        report_date,
        client_hash_id,
        content_hash_id,
        COUNT(*) AS row_count
    FROM read_parquet('{march_file}')
    GROUP BY
        report_date,
        client_hash_id,
        content_hash_id
    HAVING COUNT(*) > 1
)
""").fetchdf()

print("VERIFICATION 1 — GRAIN")
print(grain_check)


# Verification 2 — Row count and date span
count_date_check = con.execute(f"""
SELECT
    COUNT(*) AS row_count,
    MIN(report_date) AS first_date,
    MAX(report_date) AS last_date
FROM read_parquet('{march_file}')
""").fetchdf()

print("\nVERIFICATION 2 — ROW COUNT AND DATE SPAN")
print(count_date_check)


# Verification 3 — Availability
availability_check = con.execute(f"""
SELECT
    COUNT(*) AS total_rows,
    COUNT(*) FILTER (
        WHERE gsc_data_available IS TRUE
    ) AS available_gsc_rows
FROM read_parquet('{march_file}')
""").fetchdf()

print("\nVERIFICATION 3 — GSC AVAILABILITY")
print(availability_check)
```

## 4. Data limits

*What can this data never tell you? Unbalanced history, GSC-only early rows, window overlaps.*

```python
print("""
DATA LIMITATION

Named limitation: incomplete measurement coverage.

The warehouse is an unbalanced panel, and GSC availability can vary by
date and client. Therefore, an unavailable GSC observation cannot safely
be interpreted as zero traffic.

This means the data cannot tell us whether zero activity represents
genuinely zero traffic or missing measurement. Therefore, the future
outcome should only be assigned when GSC data is actually available.

The ranking is also only a proxy for refresh opportunity and should be
used as decision support rather than as a causal explanation of why
content needs refreshing.
""")
```
