## 1. My lane as an ML task (type)

*Classification, clustering, ranking, or scoring — which one, and why?*

My lane is **Refresh / Content Opportunity Scoring**, which is best framed as a **ranking (or scoring)** machine learning task.

The objective is not to classify pages into fixed categories, but to assign each webpage an opportunity score and rank pages from highest to lowest priority for review. Pages with higher scores are expected to have a greater opportunity to benefit from a content refresh.

The ranked output helps content teams decide which pages should be reviewed first when resources are limited.

```python
ml_task_type = "Ranking / Scoring"

print("ML Task Type:", ml_task_type)
```

## 2. Target or proxy

*What would you predict? Where does that label come from — observed outcome or a defined rule?*

There is no direct label in the dataset that says whether a page should be refreshed. Therefore, the model would use a **proxy target**.

A suitable proxy is a **Refresh Opportunity Score**, which estimates how likely a webpage is to benefit from a content refresh based on observed search and engagement signals.

This proxy is defined from historical measurements rather than a manually labeled "needs refresh" column. The prediction is intended to support human decision making instead of replacing it.
### Content Action

The model's output is a ranked list of webpages. Content editors review the highest-ranked pages first, decide whether the content should be refreshed, update the selected pages, and later monitor whether performance improves. ## 3. Success metric


Since this is a ranking task, the most important goal is to rank the pages with the highest refresh opportunity near the top of the list.

A suitable evaluation metric is **NDCG (Normalized Discounted Cumulative Gain)** because it measures the quality of a ranked list. A higher NDCG score indicates that the pages with the greatest refresh opportunity appear earlier in the ranking.

The model is considered useful if it consistently prioritizes the pages that are most valuable for content teams to review first.

```python
metric = "NDCG (Normalized Discounted Cumulative Gain)"
print("Selected evaluation metric:", metric)
```

## Why Machine Learning Beats a Fixed Rule

A simple rule such as "refresh every page with CTR below 2%" is not sufficient because content performance depends on many interacting factors.

For example, impressions, clicks, sessions, CTR, search trends, and other performance signals all influence whether a page should be refreshed. These relationships are often complex and cannot be captured by a single threshold.

Machine learning can analyze multiple signals together and learn patterns from historical observations to produce a more useful ranking of pages. The recommendations remain decision-support tools that help content teams prioritize their work rather than replace human judgment.

```python
signals = [
    "impressions_90d",
    "sessions_90d",
    "ctr",
    "trend_direction"
]

print("Example signals that may be used for ranking:")

for signal in signals:
    print("-", signal)
```
